```python
import pandas as pd
```

```python

import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python

from pandas.plotting import parallel_coordinates
```

```python
# Load dataset
df = pd.read_excel("STTHK3033_FA_SectionA(Q1).xlsx")

print(df.shape)
print(df.columns)
```

# SPLOM

```python
selected_sp = [
    'GDP_Per_Capita_USD',
    'Life_Expectancy_At_Birth',
    'Infant_Mortality_Rate_per_1000',
    'Physicians_per_1000',
    'Current_Health_Expenditure_Per_Capita_USD'
]

sample_df = df[selected_sp].sample(500, random_state=42)

sns.pairplot(
    sample_df,
    diag_kind='hist'
)

plt.suptitle(
    "Scatterplot Matrix of Healthcare Indicators",
    y=1.02
)

plt.show()
```

# Parallel Coordinates

```python
selected_sp = [
    'GDP_Per_Capita_USD',
    'Life_Expectancy_At_Birth',
    'Infant_Mortality_Rate_per_1000',
    'Physicians_per_1000',
    'Current_Health_Expenditure_Per_Capita_USD'
]

pc_df = df[selected_sp + ['Region']].sample(
    300,
    random_state=42
)

plt.figure(figsize=(14,8))

parallel_coordinates(
    pc_df,
    class_column='Region',
    alpha=0.3
)

plt.xticks(rotation=45)
plt.title("Parallel Coordinates of Healthcare Indicators")
plt.tight_layout()
plt.show()
```
